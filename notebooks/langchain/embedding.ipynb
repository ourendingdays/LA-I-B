### Embedding models convert documents into numerical vectors. 

These vectors capture the semantic meaning of the text, enabling more accurate and context-aware search capabilities. Document embedding is a powerful technique to convert textual data into numerical vectors, which can then be used for various downstream tasks such as search, classification, clustering, and mor

```python
# All needed imports
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Loading data file
loader = TextLoader("../../data/raw/txt/rag_notebook.txt")
data = loader.load()
print(data)
```

```python
# Splitting data
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    length_function=len,
)

chunks = text_splitter.split_text(data[0].page_content)
chunks
```

```python
# Hugging Face Embedding Model
model_name = "sentence-transformers/all-mpnet-base-v2"
huggingface_embedding = HuggingFaceEmbeddings(model_name=model_name)

query = "How are you?"
query_embedding = huggingface_embedding.embed_query(query)
```

```python
query_embedding[:10] # Display the first 10 dimensions of the embedding vector
```

```python
doc_result = huggingface_embedding.embed_documents(chunks)
# see the shpe of the embedding vector
print("Length of the first document embedding vector:", len(doc_result[0]))
print("Total number of document embedding vectors:", len(doc_result))
```
