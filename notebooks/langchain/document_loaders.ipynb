```python
# All needed imports

# Different loaders. They Load data into the Document object. there are still many more
from bs4 import BeautifulSoup
from langchain_community.document_loaders import TextLoader, PyPDFLoader, PyMuPDFLoader, JSONLoader, CSVLoader, WebBaseLoader, UnstructuredFileLoader, Docx2txtLoader, ArxivLoader
from langchain_core.documents import Document
import json
from pathlib import Path
from pprint import pprint
import requests
```

```python
from langchain_core.documents import Document

d = Document(page_content="""Python is an interpreted high-level general-purpose programming language. 
                        Python's design philosophy emphasizes code readability with its notable use of significant indentation.""",
         metadata={
             'my_document_id' : 234234,
             'my_document_source' : "About Python",
             'my_document_create_time' : 1680013019
         })

pprint(d)
```

```python
# Loading text file
loader = TextLoader("../../data/raw/txt/rag_notebook.txt")
data = loader.load()
print(data)
```

```python
# Loading PDF file
loader = PyPDFLoader("../../data/raw/pdf/Full-48.pdf")
data = loader.load()
print(data)
```

```python
# Loading PDf file with PyMuPDFLoader - fastest PDF loader in LangChain and has more metadata, than PyPDFLoader
loader = PyMuPDFLoader("../../data/raw/pdf/Full-48.pdf")
data = loader.load_and_split()
print(data)
```

```python
# Loading json file
data = json.loads(Path("../../data/raw/example_json.json").read_text())

print(data)
```

```python
# Load a webpage and parse it with BeautifulSoup
url = "https://www.ibm.com/de-de/think/tutorials/build-multimodal-rag-langchain-with-docling-granite?regionCode=de&languageCode=de&cm-history=de-de"
response = requests.get(url)

soup = BeautifulSoup(response.content, "html.parser")

pprint(soup.prettify())
```

```python
# Loading webpage with WebBaseLoader - this wil lcapture only text from the webpage, but not the images or other media, links, etc
loader = WebBaseLoader(url)
data = loader.load()
pprint(data)
```
